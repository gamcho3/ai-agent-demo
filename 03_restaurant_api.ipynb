api데이터 형태의 타입 설정하기

```typescript
type Restaurant = {
  OPENDATA_ID: string;
  BZ_NM: string;
  GNG_CS: string;
  FD_CS: string;
  TLNO: string;
  MBZ_HR: string;
  SEAT_CNT: string;
  PKPL: string;
  HP: string;
  PSB_FRN: string;
  BKN_YN: string;
  INFN_FCL: string;
  BRFT_YN: string;
  DSSRT_YN: string;
  MNU: string;
  SMPL_DESC: string;
  SBW: string;
  BUS: string;
};
```

```typescript
// URL주소 만들기
const url = new URL("https://www.daegufood.go.kr/kor/api/tasty.html?mode=json")
url.searchParams.set("addr","중구")
console.log(url.toString());
```

```typescript
//api호출
const response = await fetch(
  url.toString()
);

if (!response.ok) {
  throw new Error(
    `API 요청 실패: ${response.status}`
  );
}

const data = await response.json();
console.log(data)
```

```typescript
const restaurants =
  data.data as Restaurant[];

  console.log(restaurants.length)
```

```typescript
//api 호출함수 생성
async function fetchRestaurants(
  district: string
): Promise<Restaurant[]> {

  const url = new URL(
    "https://www.daegufood.go.kr/kor/api/tasty.html?mode=json",
  );

  url.searchParams.set(
    "addr",
    district
  );

  const response =
    await fetch(
      url.toString()
    );

  if (!response.ok) {

    throw new Error(
      `API 요청 실패: ${response.status}`
    );

  }

  const data =
    await response.json();

  // 실제 API 응답 구조에 맞게 수정
  return data.data as Restaurant[];
}
```

```typescript
const restaurants = await fetchRestaurants("남구");

console.log(restaurants.length);
```

카테고리 필터함수 만들기

```typescript
function filterByCategory(
  restaurants: Restaurant[],
  category: RestaurantCategory | null,
): Restaurant[] {
  if (category === null) {
    return restaurants;
  }

  return restaurants.filter((restaurant) => restaurant.FD_CS === category);
}
```

```typescript
const koreanRestaurants = filterByCategory(restaurants, "중식");

console.log(koreanRestaurants.map((restaurant) => restaurant.BZ_NM));
```
